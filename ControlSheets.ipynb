```python
Reg= pd.read_csv('Reg13-1.csv', sep=',')
```

```python
Reg.sort_values("CIC ID", inplace=True)
```

```python
CourseCode= Reg['Course Code'].unique().tolist()
Reg['CourseName']= Reg['Course Code']+'_'+Reg['Course Name']
```

```python
LA= Reg['CourseName'].unique().tolist()
```

```python
CS = dict()
for k, v in Reg.groupby('Course Code'):
    CS[k] = v
```

```python
os.chdir(r'E:\Work CIC\04 Fall 2021\CONTROL\Template\CC')
template_file= r'E:\Work CIC\04 Fall 2021\CONTROL\Template\template.xlsx'
for i in range(0, len(CourseCode)):
    filename= LA[i]
    writer= pd.ExcelWriter
    globals()["{}".format(CourseCode[i])]= CS['{}'.format(CourseCode[i])]
    wb= load_workbook(template_file)
    ws= wb.get_sheet_by_name('2012 (new)')
    rows= dataframe_to_rows(globals()["{}".format(CourseCode[i])], index= False, header= 5)
    for r_idx, row in enumerate(rows,1):
        for c_idx, value in enumerate(row,1):
            ws.cell(row=r_idx+5,column= c_idx+1, value= value)
    wb.save(filename+'.xlsx')
```
